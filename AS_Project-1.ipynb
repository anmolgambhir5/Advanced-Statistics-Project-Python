```python
# 3.1
mu=5
sigma=1.5
x=3.17
plt.figure(figsize=(6,2))
plt.axvline(x=3.17, c='r')
x = np.linspace(mu -3*sigma, mu+3*sigma,100 )
y = stats.norm.pdf(x,mu,sigma)
plt.plot(x,y, color='blue')
plt.fill_between(x,y,where=(x<3.17), color='lightblue')
plt.xlabel('breaking strength')
plt.ylabel('probability')
plt.title('Normal Distribution')
plt.show
```

```python
# for n way ANOVA
from statsmodels.formula.api import ols 
from statsmodels.stats.anova import _get_covariance, anova_lm
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
#7.1 Effect of Dentist and Method on Response
formula = 'Response ~ C(Dentist)'
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
print(aov_table)
```

```python
# 7.3
formula = 'Response ~ C (Dentist)'
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
tukey_results = pairwise_tukeyhsd(df2['Response'], df2['Dentist'])
print(aov_table)
print(tukey_results)
```

```python
plt.figure(figsize = (10,4))
sns.pointplot(x='Dentist', y='Response', data=df2, hue='Alloy', ci = None)
```

```python
# 7.4
formula = 'Response ~ C (Method)'
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
tukey_results = pairwise_tukeyhsd(df2['Response'], df2['Method'])
print(aov_table)
print(tukey_results)
```

```python
plt.figure(figsize = (10,4))
sns.pointplot(x='Method', y='Response', data=df2, hue='Alloy', ci = None)
```

```python
# 7.5 Effect of difference in Temperature levets on implant hardness separately for two types of alloys
formula = 'Response ~ Temp'
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
tukey_results = pairwise_tukeyhsd(df2['Response'], df2['Temp'])
print(aov_table)
print(tukey_results)
```

```python
# 7.5 Plotting the effect of Temperature on Implant Hardness for both types of Alloys
plt.figure(figsize = (10,4))
sns.pointplot(x='Temp', y='Response', data=df2, hue='Alloy', ci = None)
```

```python
7.6 Interaction effect of Dentist and Method on Response
formula = 'Response ~ C (Dentist) : C(Method)'
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
print(aov_table)
```

```python
# 7.6 Interaction Plot for Dentist and Response for Two types of Alloys
plt.figure(figsize = (10,4))
sns.pointplot(x='Dentist', y='Response', data=df2, hue='Alloy', ci = None)
```

```python
# 7.6 Interaction Plot for Method and Response for Two types of Alloys
plt.figure(figsize = (20,8))
sns.pointplot(x='Method', y='Response', data=df2, hue='Alloy', ci = None)
```

```python
# 7.7
formula = 'Response ~ C (Dentist)+ C (Method) : C(Alloy)' 
model = ols(formula,df2).fit()
aov_table = anova_lm(model, type=2)
tukey_results = pairwise_tukeyhsd(df2['Response'], df2['Dentist'])
print(aov_table)
print(tukey_results)
```
